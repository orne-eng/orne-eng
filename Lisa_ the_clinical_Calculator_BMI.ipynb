```python
#Sources: 
#https://www.cdc.gov/bmi/adult-calculator/index.html
#https://www.cdc.gov/bmi/adult-calculator/bmi-categories.html

Name = input ("Hello, enter your name to start")
#given that str can't be calculated remember to put int at the beginig of Weight and Height to be able to calculate. 
#silly but usual mistake there. 
Weight_Pounds = int(input("Enter your Weight in pounds."))
Height_Inches = int(input("Now enter your height in inches: "))

# weight (pounds)/[height (in)]2 x 703
BMI = ((Weight_Pounds  / (Height_Inches * Height_Inches) ) * 703)
print(BMI)      

#Parameters:

'''BMI Category
BMI Range (kg/m2)
Underweight
Less than 18.5
Healthy Weight
18.5 to less than 25
Overweight
25 to less than 30
Obesity
30 or greater
Class 1 Obesity
30 to less than 35
Class 2 Obesity
35 to less than 40
Class 3 Obesity
(Severe Obesity)
40 or greater '''

if BMI > 0:
    if (BMI < 18.5 ):
        print (Name+", you are underweigt. This answer relies strictly on the data provided. ")
    elif 18.5 <= BMI < 25:
        print (Name+",you are in a healthy weight. This answer relies strictly on the data provided. ")
    elif 25 <= BMI < 30:
        print (Name+",you are overweight. This answer relies strictly on the data provided.")
    elif 30 <= BMI < 35:
         print (Name+",you are obesse (class 1). This answer relies strictly on the data provided.")
    elif 35 <= BMI < 40:
             print (Name+",you are obesse (class 2). This answer relies strictly on the data provided.")
    elif BMI < 40:
             print (Name+",you are obesse (class 3). This answer relies strictly on the data provided.")
         
else: 
    print ("Verify the information given and try again please.")
    
```
