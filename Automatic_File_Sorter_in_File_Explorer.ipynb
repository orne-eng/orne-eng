```python
import os, shutil
#put the r in front of the path, so it read it well
#put a / at the end of the path. Each one. 
path = r"C:\Users\Ornel\Files To Sorter/"
file_name = os.listdir(path)

folder_names = ['xlsx files','JPG files', 'txt files']

for loop in range (0, 3):
    if not os.path.exists(path + folder_names[loop]):
     #print(path + folder_names[loop]) use it at first to check. 
        os.markedirs(path + folder_names[loop])

for file in file_name:
    if '.xlsx' in file and not os.path.exists(path + 'xlsx files/' + file): #/ at the end of the path. 
        shutil.move(path + file, path + 'xlsx files/' + file)
    elif '.jpg' in file and not os.path.exists(path + 'JPG files/' + file):
        shutil.move(path + file, path + 'JPG files/' + file)
    elif '.txt' in file and not os.path.exists(path + 'txt files/' + file):
        shutil.move(path + file, path + 'txt files/' + file)
else:
    print("An error ocurred? if so, please try again, but first check that the file is not already copied.")
```
